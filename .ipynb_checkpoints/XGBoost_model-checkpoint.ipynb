```python
import pandas as pd
import numpy as np
import xgboost as xgb

print("XGBoost version:", xgb.__version__)
print("Pandas version:", pd.__version__)
print("NumPy version:", np.__version__)
```

```python
# Load the modelling dataset

data = pd.read_csv("modelling_data.csv")

print("Dataset shape:", data.shape)
print("\nColumns:")
print(data.columns.tolist())

print("\nPresence distribution:")
print(data["Presence"].value_counts())

print("\nYears:")
print(sorted(data["year"].unique()))
```

```python
# Define predictors for XGBoost
# Same reduced predictor set used in the final Random Forest model

reduced_predictors = [
    "SOC",
    "Aspect",
    "Slope",
    "PH",
    "TRI",
    "TWI",
    "NDVI",
    "PCI",
    "LST",
    "pop",
    "rain",
    "LULC"
]

target = "Presence"

print("Predictors:")
for i, predictor in enumerate(reduced_predictors, 1):
    print(f"{i}. {predictor}")

print("\nNumber of predictors:", len(reduced_predictors))
print("Target:", target)
```

```python
from sklearn.model_selection import GroupShuffleSplit

# Spatial grouping by OBJECTID
groups = data["OBJECTID"]

# Create a 75% training / 25% validation spatial split
gss = GroupShuffleSplit(
    n_splits=1,
    test_size=0.25,
    random_state=42
)

train_idx, test_idx = next(
    gss.split(data, data[target], groups=groups)
)

train_data = data.iloc[train_idx].copy()
test_data = data.iloc[test_idx].copy()

print("Training rows:", len(train_data))
print("Validation rows:", len(test_data))

print("\nTraining locations:", train_data["OBJECTID"].nunique())
print("Validation locations:", test_data["OBJECTID"].nunique())

print("\nTraining presence/background:")
print(train_data[target].value_counts())

print("\nValidation presence/background:")
print(test_data[target].value_counts())

# Check that no spatial locations occur in both datasets
overlap = set(train_data["OBJECTID"]) & set(test_data["OBJECTID"])

print("\nSpatial overlap:", len(overlap))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder

# Separate numeric and categorical predictors
numeric_predictors = [
    "SOC",
    "Aspect",
    "Slope",
    "PH",
    "TRI",
    "TWI",
    "NDVI",
    "PCI",
    "LST",
    "pop",
    "rain"
]

categorical_predictors = ["LULC"]

# Preprocessing pipeline
xgb_preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            Pipeline([
                ("imputer", SimpleImputer(strategy="median"))
            ]),
            numeric_predictors
        ),
        (
            "cat",
            Pipeline([
                ("imputer", SimpleImputer(strategy="most_frequent")),
                ("onehot", OneHotEncoder(handle_unknown="ignore"))
            ]),
            categorical_predictors
        )
    ]
)

# Prepare X and y
X_train_raw = train_data[reduced_predictors]
X_test_raw = test_data[reduced_predictors]

y_train = train_data[target]
y_test = test_data[target]

# Fit preprocessing ONLY on training data
X_train = xgb_preprocessor.fit_transform(X_train_raw)
X_test = xgb_preprocessor.transform(X_test_raw)

print("Training matrix shape:", X_train.shape)
print("Validation matrix shape:", X_test.shape)
print("Training target shape:", y_train.shape)
print("Validation target shape:", y_test.shape)
```

```python
# Calculate class imbalance weight for XGBoost

negative_count = (y_train == 0).sum()
positive_count = (y_train == 1).sum()

scale_pos_weight = negative_count / positive_count

print("Background samples:", negative_count)
print("Presence samples:", positive_count)
print("scale_pos_weight:", round(scale_pos_weight, 4))
```

```python
# Train baseline XGBoost model

xgb_model = xgb.XGBClassifier(
    n_estimators=500,
    max_depth=6,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    min_child_weight=5,
    reg_lambda=1.0,
    objective="binary:logistic",
    eval_metric="logloss",
    scale_pos_weight=scale_pos_weight,
    random_state=42,
    n_jobs=-1
)

xgb_model.fit(
    X_train,
    y_train
)

print("Baseline XGBoost model trained successfully.")
```

```python
from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    accuracy_score,
    confusion_matrix,
    mean_squared_error,
    mean_absolute_error,
    r2_score
)

# Predict probability of pollinator presence
y_prob_xgb = xgb_model.predict_proba(X_test)[:, 1]

# Convert probabilities to binary predictions
y_pred_xgb = (y_prob_xgb >= 0.50).astype(int)

# Primary metrics
roc_auc_xgb = roc_auc_score(y_test, y_prob_xgb)
pr_auc_xgb = average_precision_score(y_test, y_prob_xgb)

# Classification metrics
accuracy_xgb = accuracy_score(y_test, y_pred_xgb)

tn, fp, fn, tp = confusion_matrix(
    y_test,
    y_pred_xgb
).ravel()

sensitivity_xgb = tp / (tp + fn) if (tp + fn) > 0 else np.nan
specificity_xgb = tn / (tn + fp) if (tn + fp) > 0 else np.nan

# Probability-score diagnostics
rmse_xgb = np.sqrt(mean_squared_error(y_test, y_prob_xgb))
mae_xgb = mean_absolute_error(y_test, y_prob_xgb)
r2_xgb = r2_score(y_test, y_prob_xgb)

print("BASELINE XGBOOST PERFORMANCE")
print("=" * 40)
print(f"ROC-AUC:      {roc_auc_xgb:.4f}")
print(f"PR-AUC:       {pr_auc_xgb:.4f}")
print(f"Accuracy:     {accuracy_xgb:.4f}")
print(f"Sensitivity:  {sensitivity_xgb:.4f}")
print(f"Specificity:  {specificity_xgb:.4f}")
print(f"RMSE:         {rmse_xgb:.4f}")
print(f"MAE:          {mae_xgb:.4f}")
print(f"R²:           {r2_xgb:.4f}")

print("\nConfusion Matrix")
print(confusion_matrix(y_test, y_pred_xgb))
```

```python
from sklearn.metrics import roc_auc_score, average_precision_score

# XGBoost tuning configurations
xgb_configs = {
    "Baseline": {
        "max_depth": 6,
        "learning_rate": 0.05,
        "min_child_weight": 5,
        "subsample": 0.8,
        "colsample_bytree": 0.8
    },
    
    "Shallower Trees": {
        "max_depth": 3,
        "learning_rate": 0.05,
        "min_child_weight": 5,
        "subsample": 0.8,
        "colsample_bytree": 0.8
    },
    
    "Deeper Trees": {
        "max_depth": 8,
        "learning_rate": 0.05,
        "min_child_weight": 5,
        "subsample": 0.8,
        "colsample_bytree": 0.8
    },
    
    "Higher Min Child": {
        "max_depth": 6,
        "learning_rate": 0.05,
        "min_child_weight": 10,
        "subsample": 0.8,
        "colsample_bytree": 0.8
    },
    
    "Lower Learning Rate": {
        "max_depth": 6,
        "learning_rate": 0.02,
        "min_child_weight": 5,
        "subsample": 0.8,
        "colsample_bytree": 0.8
    },
    
    "More Conservative": {
        "max_depth": 4,
        "learning_rate": 0.03,
        "min_child_weight": 10,
        "subsample": 0.8,
        "colsample_bytree": 0.8
    }
}

xgb_tuning_results = []

for name, params in xgb_configs.items():
    
    model = xgb.XGBClassifier(
        n_estimators=500,
        max_depth=params["max_depth"],
        learning_rate=params["learning_rate"],
        min_child_weight=params["min_child_weight"],
        subsample=params["subsample"],
        colsample_bytree=params["colsample_bytree"],
        reg_lambda=1.0,
        objective="binary:logistic",
        eval_metric="logloss",
        scale_pos_weight=scale_pos_weight,
        random_state=42,
        n_jobs=-1
    )
    
    model.fit(X_train, y_train)
    
    probabilities = model.predict_proba(X_test)[:, 1]
    
    roc_auc = roc_auc_score(y_test, probabilities)
    pr_auc = average_precision_score(y_test, probabilities)
    
    xgb_tuning_results.append({
        "Model": name,
        "ROC_AUC": roc_auc,
        "PR_AUC": pr_auc
    })

xgb_tuning_results_df = pd.DataFrame(xgb_tuning_results)

xgb_tuning_results_df = xgb_tuning_results_df.sort_values(
    by=["PR_AUC", "ROC_AUC"],
    ascending=False
).reset_index(drop=True)

print(xgb_tuning_results_df.to_string(index=False))
```

```python
# Best XGBoost configuration from the tuning stage

best_xgb_params = {
    "n_estimators": 500,
    "max_depth": 6,
    "learning_rate": 0.02,
    "min_child_weight": 5,
    "subsample": 0.8,
    "colsample_bytree": 0.8,
    "reg_lambda": 1.0,
    "scale_pos_weight": scale_pos_weight,
    "objective": "binary:logistic",
    "eval_metric": "logloss",
    "random_state": 42,
    "n_jobs": -1
}

print("Selected XGBoost configuration:")
for key, value in best_xgb_params.items():
    print(f"{key}: {value}")
```

```python
from sklearn.model_selection import GroupKFold
from sklearn.metrics import roc_auc_score, average_precision_score

# 5-fold spatial cross-validation
group_kfold = GroupKFold(n_splits=5)

spatial_cv_results = []

for fold, (train_idx_cv, val_idx_cv) in enumerate(
    group_kfold.split(data, data[target], groups=data["OBJECTID"]),
    start=1
):
    
    # Split raw data
    train_fold = data.iloc[train_idx_cv]
    val_fold = data.iloc[val_idx_cv]
    
    X_train_fold_raw = train_fold[reduced_predictors]
    X_val_fold_raw = val_fold[reduced_predictors]
    
    y_train_fold = train_fold[target]
    y_val_fold = val_fold[target]
    
    # Fit preprocessing ONLY on the current training fold
    fold_preprocessor = ColumnTransformer(
        transformers=[
            (
                "num",
                Pipeline([
                    ("imputer", SimpleImputer(strategy="median"))
                ]),
                numeric_predictors
            ),
            (
                "cat",
                Pipeline([
                    ("imputer", SimpleImputer(strategy="most_frequent")),
                    ("onehot", OneHotEncoder(handle_unknown="ignore"))
                ]),
                categorical_predictors
            )
        ]
    )
    
    X_train_fold = fold_preprocessor.fit_transform(X_train_fold_raw)
    X_val_fold = fold_preprocessor.transform(X_val_fold_raw)
    
    # Calculate fold-specific class weight
    fold_negative = (y_train_fold == 0).sum()
    fold_positive = (y_train_fold == 1).sum()
    fold_weight = fold_negative / fold_positive
    
    # Train XGBoost
    fold_model = xgb.XGBClassifier(
        n_estimators=500,
        max_depth=6,
        learning_rate=0.02,
        min_child_weight=5,
        subsample=0.8,
        colsample_bytree=0.8,
        reg_lambda=1.0,
        scale_pos_weight=fold_weight,
        objective="binary:logistic",
        eval_metric="logloss",
        random_state=42,
        n_jobs=-1
    )
    
    fold_model.fit(
        X_train_fold,
        y_train_fold
    )
    
    # Predict probabilities
    fold_prob = fold_model.predict_proba(X_val_fold)[:, 1]
    
    # Evaluate
    fold_roc_auc = roc_auc_score(y_val_fold, fold_prob)
    fold_pr_auc = average_precision_score(y_val_fold, fold_prob)
    
    spatial_cv_results.append({
        "Fold": fold,
        "ROC_AUC": fold_roc_auc,
        "PR_AUC": fold_pr_auc
    })
    
    print(
        f"Fold {fold}: "
        f"ROC-AUC = {fold_roc_auc:.4f}, "
        f"PR-AUC = {fold_pr_auc:.4f}"
    )

# Results table
xgb_spatial_cv_df = pd.DataFrame(spatial_cv_results)

print("\nXGBoost Spatial Cross-Validation Results")
print("=" * 45)
print(xgb_spatial_cv_df.to_string(index=False))

print("\nMean ± SD")
print(
    f"ROC-AUC: "
    f"{xgb_spatial_cv_df['ROC_AUC'].mean():.4f} ± "
    f"{xgb_spatial_cv_df['ROC_AUC'].std():.4f}"
)

print(
    f"PR-AUC:  "
    f"{xgb_spatial_cv_df['PR_AUC'].mean():.4f} ± "
    f"{xgb_spatial_cv_df['PR_AUC'].std():.4f}"
)
```

```python
# Train the final XGBoost model using the selected configuration

final_xgb = xgb.XGBClassifier(
    n_estimators=500,
    max_depth=6,
    learning_rate=0.02,
    min_child_weight=5,
    subsample=0.8,
    colsample_bytree=0.8,
    reg_lambda=1.0,
    objective="binary:logistic",
    eval_metric="logloss",
    scale_pos_weight=scale_pos_weight,
    random_state=42,
    n_jobs=-1
)

final_xgb.fit(
    X_train,
    y_train
)

print("Final XGBoost model trained successfully.")
```

```python
# Final independent spatial holdout evaluation

from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    accuracy_score,
    confusion_matrix
)

# Predict probability of pollinator presence
final_xgb_prob = final_xgb.predict_proba(X_test)[:, 1]

# Binary predictions at 0.50 threshold
final_xgb_pred = (final_xgb_prob >= 0.50).astype(int)

# Primary performance metrics
final_xgb_roc_auc = roc_auc_score(
    y_test,
    final_xgb_prob
)

final_xgb_pr_auc = average_precision_score(
    y_test,
    final_xgb_prob
)

# Classification metrics
final_xgb_accuracy = accuracy_score(
    y_test,
    final_xgb_pred
)

tn, fp, fn, tp = confusion_matrix(
    y_test,
    final_xgb_pred
).ravel()

final_xgb_sensitivity = tp / (tp + fn)
final_xgb_specificity = tn / (tn + fp)

print("FINAL XGBOOST SPATIAL HOLDOUT PERFORMANCE")
print("=" * 50)

print(f"ROC-AUC:      {final_xgb_roc_auc:.4f}")
print(f"PR-AUC:       {final_xgb_pr_auc:.4f}")
print(f"Accuracy:     {final_xgb_accuracy:.4f}")
print(f"Sensitivity:  {final_xgb_sensitivity:.4f}")
print(f"Specificity:  {final_xgb_specificity:.4f}")

print("\nConfusion Matrix")
print(confusion_matrix(y_test, final_xgb_pred))
```

```python
from pathlib import Path
import rasterio

aligned_dir = Path(r"D:\CODING\Modelling\aligned_predictors")

years = [1995, 2005, 2015, 2025]

predictor_names = [
    "SOC",
    "Aspect",
    "Slope",
    "PH",
    "TRI",
    "TWI",
    "NDVI",
    "PCI",
    "LST",
    "pop",
    "rain",
    "LULC"
]

print("VERIFYING ALIGNED RASTER GRIDS FOR ALL YEARS")
print("=" * 70)

reference = None
all_match = True

for year in years:
    print(f"\nYear: {year}")
    print("-" * 40)
    
    for predictor in predictor_names:
        filename = f"{predictor}_{year}_aligned.tif"
        path = aligned_dir / filename
        
        if not path.exists():
            print(f"✗ MISSING: {filename}")
            all_match = False
            continue
        
        with rasterio.open(path) as src:
            current = {
                "shape": (src.height, src.width),
                "crs": src.crs,
                "transform": src.transform
            }
            
            if reference is None:
                reference = current
                print(f"Reference grid established from {filename}")
            else:
                matches = (
                    current["shape"] == reference["shape"]
                    and current["crs"] == reference["crs"]
                    and current["transform"] == reference["transform"]
                )
                
                if not matches:
                    print(f"✗ GRID MISMATCH: {filename}")
                    all_match = False

print("\n" + "=" * 70)

if all_match:
    print("✓ ALL 48 ALIGNED RASTERS HAVE THE SAME GRID.")
    print("Ready for XGBoost PHSI raster generation.")
else:
    print("✗ GRID PROBLEMS FOUND — DO NOT PROCEED TO MAPPING YET.")
```

```python
from pathlib import Path
import numpy as np
import pandas as pd
import rasterio

# ------------------------------------------------------------
# SETTINGS
# ------------------------------------------------------------

base_dir = Path(r"D:\CODING\Modelling")
aligned_dir = base_dir / "aligned_predictors"
output_dir = base_dir / "XGB_PHSI_outputs"

output_dir.mkdir(exist_ok=True)

year = 1995

predictors = [
    "SOC",
    "Aspect",
    "Slope",
    "PH",
    "TRI",
    "TWI",
    "NDVI",
    "PCI",
    "LST",
    "pop",
    "rain",
    "LULC"
]

# ------------------------------------------------------------
# RASTER FILES
# ------------------------------------------------------------

raster_paths = {
    predictor: aligned_dir / f"{predictor}_{year}_aligned.tif"
    for predictor in predictors
}

# ------------------------------------------------------------
# OPEN REFERENCE RASTER
# ------------------------------------------------------------

reference_path = raster_paths["SOC"]

with rasterio.open(reference_path) as ref:
    profile = ref.profile.copy()
    height = ref.height
    width = ref.width
    transform = ref.transform
    crs = ref.crs
    nodata = ref.nodata

print("Generating XGBoost PHSI")
print("=" * 60)
print(f"Year: {year}")
print(f"Raster size: {width} × {height}")
print(f"Total pixels: {width * height:,}")
print(f"CRS: {crs}")

# ------------------------------------------------------------
# OUTPUT PROFILE
# ------------------------------------------------------------

output_profile = profile.copy()

output_profile.update({
    "driver": "GTiff",
    "dtype": "float32",
    "count": 1,
    "nodata": -9999.0,
    "compress": "deflate",
    "predictor": 2
})

output_path = output_dir / f"PHSI_XGB_{year}.tif"

# ------------------------------------------------------------
# PROCESS BY ROW BLOCKS
# ------------------------------------------------------------

block_height = 128

total_valid = 0
total_nodata = 0

with rasterio.open(output_path, "w", **output_profile) as dst:

    # Open all predictor rasters
    sources = {
        predictor: rasterio.open(path)
        for predictor, path in raster_paths.items()
    }

    try:

        for row_start in range(0, height, block_height):

            rows = min(block_height, height - row_start)

            # ------------------------------------------------
            # READ PREDICTOR BLOCKS
            # ------------------------------------------------

            arrays = {}

            for predictor, src in sources.items():

                arr = src.read(
                    1,
                    window=rasterio.windows.Window(
                        0,
                        row_start,
                        width,
                        rows
                    )
                ).astype(np.float32)

                arrays[predictor] = arr

            # ------------------------------------------------
            # BUILD VALID PIXEL MASK
            # ------------------------------------------------

            valid_mask = np.ones(
                (rows, width),
                dtype=bool
            )

            for predictor, arr in arrays.items():

                if predictor == "LULC":

                    # LULC must contain valid class values
                    valid_mask &= np.isfinite(arr)
                    valid_mask &= np.isin(
                        arr,
                        [0, 1, 2, 3, 4]
                    )

                else:

                    valid_mask &= np.isfinite(arr)

            valid_rows, valid_cols = np.where(valid_mask)

            # Output block initialized as NoData
            output_block = np.full(
                (rows, width),
                -9999.0,
                dtype=np.float32
            )

            if len(valid_rows) > 0:

                # --------------------------------------------
                # CREATE MODEL INPUT
                # --------------------------------------------

                model_data = pd.DataFrame({
                    "SOC": arrays["SOC"][valid_rows, valid_cols],
                    "Aspect": arrays["Aspect"][valid_rows, valid_cols],
                    "Slope": arrays["Slope"][valid_rows, valid_cols],
                    "PH": arrays["PH"][valid_rows, valid_cols],
                    "TRI": arrays["TRI"][valid_rows, valid_cols],
                    "TWI": arrays["TWI"][valid_rows, valid_cols],
                    "NDVI": arrays["NDVI"][valid_rows, valid_cols],
                    "PCI": arrays["PCI"][valid_rows, valid_cols],
                    "LST": arrays["LST"][valid_rows, valid_cols],
                    "pop": arrays["pop"][valid_rows, valid_cols],
                    "rain": arrays["rain"][valid_rows, valid_cols],
                    "LULC": arrays["LULC"][valid_rows, valid_cols]
                })

                # --------------------------------------------
                # APPLY SAME PREPROCESSING AS TRAINING
                # --------------------------------------------

                X_block = xgb_preprocessor.transform(
                    model_data
                )

                # --------------------------------------------
                # XGBOOST PREDICTIONS
                # --------------------------------------------

                probabilities = final_xgb.predict_proba(
                    X_block
                )[:, 1]

                probabilities = np.clip(
                    probabilities,
                    0,
                    1
                ).astype(np.float32)

                # --------------------------------------------
                # PLACE PREDICTIONS BACK INTO RASTER
                # --------------------------------------------

                output_block[
                    valid_rows,
                    valid_cols
                ] = probabilities

                total_valid += len(probabilities)

            total_nodata += np.sum(~valid_mask)

            # Write block
            dst.write(
                output_block,
                1,
                window=rasterio.windows.Window(
                    0,
                    row_start,
                    width,
                    rows
                )
            )

            # Progress
            if row_start % (block_height * 5) == 0:
                progress = (
                    (row_start + rows) / height
                ) * 100

                print(
                    f"Progress: {progress:.1f}%"
                )

    finally:

        for src in sources.values():
            src.close()

print("\n" + "=" * 60)
print("XGBoost PHSI generation completed.")
print(f"Output: {output_path}")
print(f"Valid pixels: {total_valid:,}")
print(f"NoData pixels: {total_nodata:,}")
```

```python
from pathlib import Path
import numpy as np
import pandas as pd
import rasterio

# ------------------------------------------------------------
# SETTINGS
# ------------------------------------------------------------

base_dir = Path(r"D:\CODING\Modelling")
aligned_dir = base_dir / "aligned_predictors"
output_dir = base_dir / "XGB_PHSI_outputs"

output_dir.mkdir(exist_ok=True)

years_to_process = [2005, 2015, 2025]

predictors = [
    "SOC",
    "Aspect",
    "Slope",
    "PH",
    "TRI",
    "TWI",
    "NDVI",
    "PCI",
    "LST",
    "pop",
    "rain",
    "LULC"
]

block_height = 128

# ------------------------------------------------------------
# PROCESS EACH YEAR
# ------------------------------------------------------------

for year in years_to_process:

    print("\n" + "=" * 70)
    print(f"GENERATING XGBOOST PHSI — {year}")
    print("=" * 70)

    # Raster paths
    raster_paths = {
        predictor: aligned_dir / f"{predictor}_{year}_aligned.tif"
        for predictor in predictors
    }

    # Reference raster
    with rasterio.open(raster_paths["SOC"]) as ref:
        profile = ref.profile.copy()
        height = ref.height
        width = ref.width
        crs = ref.crs

    print(f"Raster size: {width} × {height}")
    print(f"Total pixels: {width * height:,}")
    print(f"CRS: {crs}")

    # Output profile
    output_profile = profile.copy()

    output_profile.update({
        "driver": "GTiff",
        "dtype": "float32",
        "count": 1,
        "nodata": -9999.0,
        "compress": "deflate",
        "predictor": 2
    })

    output_path = output_dir / f"PHSI_XGB_{year}.tif"

    total_valid = 0
    total_nodata = 0

    # Open all predictors
    sources = {
        predictor: rasterio.open(path)
        for predictor, path in raster_paths.items()
    }

    try:

        with rasterio.open(
            output_path,
            "w",
            **output_profile
        ) as dst:

            for row_start in range(
                0,
                height,
                block_height
            ):

                rows = min(
                    block_height,
                    height - row_start
                )

                # --------------------------------------------
                # READ BLOCKS
                # --------------------------------------------

                arrays = {}

                window = rasterio.windows.Window(
                    0,
                    row_start,
                    width,
                    rows
                )

                for predictor, src in sources.items():

                    arrays[predictor] = src.read(
                        1,
                        window=window
                    ).astype(np.float32)

                # --------------------------------------------
                # VALID PIXEL MASK
                # --------------------------------------------

                valid_mask = np.ones(
                    (rows, width),
                    dtype=bool
                )

                for predictor, arr in arrays.items():

                    valid_mask &= np.isfinite(arr)

                    if predictor == "LULC":
                        valid_mask &= np.isin(
                            arr,
                            [0, 1, 2, 3, 4]
                        )

                valid_rows, valid_cols = np.where(
                    valid_mask
                )

                output_block = np.full(
                    (rows, width),
                    -9999.0,
                    dtype=np.float32
                )

                if len(valid_rows) > 0:

                    # ----------------------------------------
                    # MODEL INPUT
                    # ----------------------------------------

                    model_data = pd.DataFrame({
                        "SOC": arrays["SOC"][
                            valid_rows, valid_cols
                        ],
                        "Aspect": arrays["Aspect"][
                            valid_rows, valid_cols
                        ],
                        "Slope": arrays["Slope"][
                            valid_rows, valid_cols
                        ],
                        "PH": arrays["PH"][
                            valid_rows, valid_cols
                        ],
                        "TRI": arrays["TRI"][
                            valid_rows, valid_cols
                        ],
                        "TWI": arrays["TWI"][
                            valid_rows, valid_cols
                        ],
                        "NDVI": arrays["NDVI"][
                            valid_rows, valid_cols
                        ],
                        "PCI": arrays["PCI"][
                            valid_rows, valid_cols
                        ],
                        "LST": arrays["LST"][
                            valid_rows, valid_cols
                        ],
                        "pop": arrays["pop"][
                            valid_rows, valid_cols
                        ],
                        "rain": arrays["rain"][
                            valid_rows, valid_cols
                        ],
                        "LULC": arrays["LULC"][
                            valid_rows, valid_cols
                        ]
                    })

                    # ----------------------------------------
                    # SAME PREPROCESSING USED IN TRAINING
                    # ----------------------------------------

                    X_block = xgb_preprocessor.transform(
                        model_data
                    )

                    # ----------------------------------------
                    # XGBOOST PREDICTION
                    # ----------------------------------------

                    probabilities = final_xgb.predict_proba(
                        X_block
                    )[:, 1]

                    probabilities = np.clip(
                        probabilities,
                        0,
                        1
                    ).astype(np.float32)

                    # ----------------------------------------
                    # WRITE PREDICTIONS
                    # ----------------------------------------

                    output_block[
                        valid_rows,
                        valid_cols
                    ] = probabilities

                    total_valid += len(probabilities)

                total_nodata += np.sum(
                    ~valid_mask
                )

                dst.write(
                    output_block,
                    1,
                    window=window
                )

                # Progress
                if row_start % (block_height * 5) == 0:

                    progress = (
                        (row_start + rows)
                        / height
                    ) * 100

                    print(
                        f"Progress: {progress:.1f}%"
                    )

    finally:

        for src in sources.values():
            src.close()

    print(f"\n✓ {year} completed")
    print(f"Valid pixels: {total_valid:,}")
    print(f"NoData pixels: {total_nodata:,}")
    print(f"Output: {output_path}")

print("\n" + "=" * 70)
print("ALL REMAINING XGBOOST PHSI MAPS GENERATED.")
print("=" * 70)
```

```python
from pathlib import Path
import numpy as np
import rasterio

xgb_dir = Path(
    r"D:\CODING\Modelling\XGB_PHSI_outputs"
)

years = [1995, 2005, 2015, 2025]

print("XGBOOST PHSI — FOUR-YEAR QUALITY CONTROL")
print("=" * 75)

for year in years:

    path = xgb_dir / f"PHSI_XGB_{year}.tif"

    with rasterio.open(path) as src:

        phsi = src.read(1)
        valid = phsi != src.nodata
        values = phsi[valid]

        print(f"\n{year}")
        print("-" * 40)

        print(f"Valid pixels: {len(values):,}")
        print(f"NoData pixels: {(~valid).sum():,}")
        print(f"Minimum: {values.min():.6f}")
        print(f"Maximum: {values.max():.6f}")
        print(f"Mean:    {values.mean():.6f}")
        print(f"Median:  {np.median(values):.6f}")

        range_ok = (
            values.min() >= 0
            and values.max() <= 1
        )

        finite_ok = np.isfinite(values).all()

        print(
            "Range 0–1: ",
            "✓ PASS" if range_ok else "✗ FAIL"
        )

        print(
            "Finite values:",
            "✓ PASS" if finite_ok else "✗ FAIL"
        )

print("\n" + "=" * 75)
print("XGBoost PHSI quality control complete.")
```

```python
from sklearn.inspection import permutation_importance
import pandas as pd
import matplotlib.pyplot as plt

# Permutation importance on the independent spatial holdout
perm_result = permutation_importance(
    final_xgb,
    X_test,
    y_test,
    scoring="average_precision",
    n_repeats=10,
    random_state=42,
    n_jobs=-1
)

# Get transformed feature names
feature_names = xgb_preprocessor.get_feature_names_out()

# Create results table
xgb_perm_importance = pd.DataFrame({
    "Feature": feature_names,
    "Importance_Mean": perm_result.importances_mean,
    "Importance_SD": perm_result.importances_std
})

# Sort from most to least important
xgb_perm_importance = xgb_perm_importance.sort_values(
    "Importance_Mean",
    ascending=False
).reset_index(drop=True)

print("XGBoost Permutation Importance:")
display(xgb_perm_importance)

# Plot
plt.figure(figsize=(9, 7))

plt.barh(
    xgb_perm_importance["Feature"],
    xgb_perm_importance["Importance_Mean"],
    xerr=xgb_perm_importance["Importance_SD"]
)

plt.gca().invert_yaxis()
plt.xlabel("Decrease in Average Precision")
plt.ylabel("Predictor")
plt.title("XGBoost Permutation Importance")
plt.tight_layout()

# Save figure
plt.savefig(
    xgb_output_dir / "XGB_permutation_importance.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print(
    "\nFigure saved:",
    xgb_output_dir / "XGB_permutation_importance.png"
)
```
