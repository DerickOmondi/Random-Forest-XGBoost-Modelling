```python
from pathlib import Path
import numpy as np
import rasterio

# Main project directory
project_dir = Path(r"D:\CODING\Modelling")

# Input folders
rf_dir = project_dir / "RF_PHSI_outputs"
xgb_dir = project_dir / "XGB_PHSI_outputs"

# Output folder
combined_dir = project_dir / "Combined_PHSI_outputs"
combined_dir.mkdir(exist_ok=True)

# Years
years = [1995, 2005, 2015, 2025]

print("Project directory:", project_dir)
print("RF folder:", rf_dir)
print("XGBoost folder:", xgb_dir)
print("Combined output folder:", combined_dir)
print("Years:", years)
```

```python
# Expected PHSI files
rf_files = {
    year: rf_dir / f"PHSI_{year}.tif"
    for year in years
}

xgb_files = {
    year: xgb_dir / f"PHSI_XGB_{year}.tif"
    for year in years
}

print("RF PHSI files:")
for year, path in rf_files.items():
    print(year, "→", path.exists(), "|", path.name)

print("\nXGBoost PHSI files:")
for year, path in xgb_files.items():
    print(year, "→", path.exists(), "|", path.name)
```

```python
# Check RF and XGBoost raster alignment for every year

for year in years:
    with rasterio.open(rf_files[year]) as rf:
        with rasterio.open(xgb_files[year]) as xgb:
            
            same_shape = rf.shape == xgb.shape
            same_crs = rf.crs == xgb.crs
            same_transform = rf.transform == xgb.transform
            same_resolution = rf.res == xgb.res
            
            print(f"\n{year}")
            print("  Shape:", rf.shape, "| Match:", same_shape)
            print("  CRS:", rf.crs, "| Match:", same_crs)
            print("  Resolution:", rf.res, "| Match:", same_resolution)
            print("  Transform match:", same_transform)
            print("  RF NoData:", rf.nodata)
            print("  XGB NoData:", xgb.nodata)
```

```python
# Spatial cross-validation PR-AUC values
rf_pr_auc = 0.0761
xgb_pr_auc = 0.0837

# Normalize the PR-AUC values to obtain ensemble weights
total_pr_auc = rf_pr_auc + xgb_pr_auc

rf_weight = rf_pr_auc / total_pr_auc
xgb_weight = xgb_pr_auc / total_pr_auc

print("RF PR-AUC:", rf_pr_auc)
print("XGBoost PR-AUC:", xgb_pr_auc)

print("\nEnsemble weights:")
print(f"RF weight:      {rf_weight:.4f}")
print(f"XGBoost weight: {xgb_weight:.4f}")

print("\nWeight sum:", rf_weight + xgb_weight)
```

```python
# Create the combined PHSI for 1995

year = 1995

with rasterio.open(rf_files[year]) as rf_src:
    rf = rf_src.read(1).astype("float32")
    profile = rf_src.profile.copy()

with rasterio.open(xgb_files[year]) as xgb_src:
    xgb = xgb_src.read(1).astype("float32")

# Identify valid pixels in both models
rf_valid = np.isfinite(rf)
xgb_valid = np.isfinite(xgb) & (xgb != -9999)

# Pixels valid in both rasters
valid = rf_valid & xgb_valid

# Initialize output with NoData
combined = np.full(rf.shape, np.nan, dtype="float32")

# Weighted ensemble
combined[valid] = (
    rf_weight * rf[valid] +
    xgb_weight * xgb[valid]
)

# Output file
output_file = combined_dir / f"Combined_PHSI_{year}.tif"

# Update raster profile
profile.update(
    dtype="float32",
    count=1,
    nodata=np.nan,
    compress="deflate"
)

# Save combined raster
with rasterio.open(output_file, "w", **profile) as dst:
    dst.write(combined, 1)

# Basic statistics
valid_values = combined[np.isfinite(combined)]

print("Year:", year)
print("Output:", output_file)
print("Valid pixels:", len(valid_values))
print("NoData pixels:", np.count_nonzero(~np.isfinite(combined)))
print("Minimum:", valid_values.min())
print("Maximum:", valid_values.max())
print("Mean:", valid_values.mean())
print("Median:", np.median(valid_values))
```

```python
# Quality control for the 1995 combined PHSI

with rasterio.open(combined_dir / "Combined_PHSI_1995.tif") as src:
    combined_1995 = src.read(1)
    
    print("CRS:", src.crs)
    print("Shape:", src.shape)
    print("Resolution:", src.res)
    print("NoData:", src.nodata)
    
    valid_values = combined_1995[np.isfinite(combined_1995)]
    
    print("\nValue checks:")
    print("Minimum:", valid_values.min())
    print("Maximum:", valid_values.max())
    print("All values >= 0:", np.all(valid_values >= 0))
    print("All values <= 1:", np.all(valid_values <= 1))
    print("All values finite:", np.all(np.isfinite(valid_values)))
```

```python
# Generate combined PHSI rasters for 2005, 2015 and 2025

for year in [2005, 2015, 2025]:

    print(f"\nProcessing {year}...")

    # Read RF raster
    with rasterio.open(rf_files[year]) as rf_src:
        rf = rf_src.read(1).astype("float32")
        profile = rf_src.profile.copy()

    # Read XGBoost raster
    with rasterio.open(xgb_files[year]) as xgb_src:
        xgb = xgb_src.read(1).astype("float32")

    # Identify valid pixels
    rf_valid = np.isfinite(rf)
    xgb_valid = np.isfinite(xgb) & (xgb != -9999)

    # Pixels valid in both models
    valid = rf_valid & xgb_valid

    # Create output array
    combined = np.full(rf.shape, np.nan, dtype="float32")

    # Weighted ensemble
    combined[valid] = (
        rf_weight * rf[valid] +
        xgb_weight * xgb[valid]
    )

    # Output path
    output_file = combined_dir / f"Combined_PHSI_{year}.tif"

    # Update profile
    profile.update(
        dtype="float32",
        count=1,
        nodata=np.nan,
        compress="deflate"
    )

    # Save raster
    with rasterio.open(output_file, "w", **profile) as dst:
        dst.write(combined, 1)

    # Statistics
    valid_values = combined[np.isfinite(combined)]

    print("Output:", output_file)
    print("Valid pixels:", len(valid_values))
    print("NoData pixels:", np.count_nonzero(~np.isfinite(combined)))
    print("Minimum:", valid_values.min())
    print("Maximum:", valid_values.max())
    print("Mean:", valid_values.mean())
    print("Median:", np.median(valid_values))

print("\nAll remaining years completed.")
```

```python
# Final quality control for all combined PHSI rasters

print("FINAL COMBINED PHSI QC")
print("=" * 60)

for year in years:
    
    output_file = combined_dir / f"Combined_PHSI_{year}.tif"
    
    with rasterio.open(output_file) as src:
        data = src.read(1)
        valid_values = data[np.isfinite(data)]
        
        print(f"\n{year}")
        print("-" * 30)
        print("File exists:", output_file.exists())
        print("Shape:", src.shape)
        print("CRS:", src.crs)
        print("Resolution:", src.res)
        print("NoData:", src.nodata)
        print("Valid pixels:", len(valid_values))
        print("NoData pixels:", np.count_nonzero(~np.isfinite(data)))
        print("Minimum:", valid_values.min())
        print("Maximum:", valid_values.max())
        print("Mean:", valid_values.mean())
        print("Median:", np.median(valid_values))
        print("All values >= 0:", np.all(valid_values >= 0))
        print("All values <= 1:", np.all(valid_values <= 1))
        print("All values finite:", np.all(np.isfinite(valid_values)))
```

```python
# ============================================================
# CELL 9 — CLASSIFY COMBINED PHSI FOR VISUALIZATION
# ============================================================

import numpy as np
import rasterio

# Classification thresholds
# 1 = Very Low
# 2 = Low
# 3 = Moderate
# 4 = High
# 5 = Very High

classified_dir = combined_dir / "Classified_PHSI"
classified_dir.mkdir(exist_ok=True)

def classify_phsi(value):
    if 0 <= value < 0.20:
        return 1
    elif 0.20 <= value < 0.40:
        return 2
    elif 0.40 <= value < 0.60:
        return 3
    elif 0.60 <= value < 0.80:
        return 4
    elif 0.80 <= value <= 1.00:
        return 5
    else:
        return 0


for year in years:

    input_file = combined_dir / f"Combined_PHSI_{year}.tif"
    output_file = classified_dir / f"Combined_PHSI_Classified_{year}.tif"

    with rasterio.open(input_file) as src:

        profile = src.profile.copy()

        # Read continuous PHSI
        phsi = src.read(1)

        # Valid pixels
        valid = np.isfinite(phsi)

        # Create classified raster
        classified = np.zeros(phsi.shape, dtype=np.uint8)

        classified[(phsi >= 0) & (phsi < 0.20)] = 1
        classified[(phsi >= 0.20) & (phsi < 0.40)] = 2
        classified[(phsi >= 0.40) & (phsi < 0.60)] = 3
        classified[(phsi >= 0.60) & (phsi < 0.80)] = 4
        classified[(phsi >= 0.80) & (phsi <= 1.00)] = 5

        # Ensure invalid pixels remain NoData
        classified[~valid] = 0

        profile.update(
            dtype="uint8",
            count=1,
            nodata=0,
            compress="DEFLATE"
        )

        with rasterio.open(output_file, "w", **profile) as dst:
            dst.write(classified, 1)

    print(f"{year}: classified raster created")
    print(f"Output: {output_file}")
```

```python
# ============================================================
# CELL 10 — QC CLASSIFIED PHSI RASTERS
# ============================================================

import numpy as np
import rasterio

class_names = {
    0: "NoData",
    1: "Very Low",
    2: "Low",
    3: "Moderate",
    4: "High",
    5: "Very High"
}

for year in years:

    classified_file = (
        classified_dir / f"Combined_PHSI_Classified_{year}.tif"
    )

    with rasterio.open(classified_file) as src:

        raster = src.read(1)

        unique, counts = np.unique(raster, return_counts=True)

        print("\n" + "=" * 55)
        print(f"{year} CLASSIFICATION QC")
        print("=" * 55)

        total_valid = 0

        for value, count in zip(unique, counts):

            name = class_names.get(int(value), "Unexpected value")

            print(
                f"Class {int(value)} ({name}): "
                f"{count:,} pixels"
            )

            if value != 0:
                total_valid += count

        print(f"Total valid pixels: {total_valid:,}")

        # Check for unexpected class values
        unexpected = unique[(unique < 0) | (unique > 5)]

        if len(unexpected) == 0:
            print("Class values check: PASS")
        else:
            print("Class values check: FAIL")
            print("Unexpected values:", unexpected)

        # Check raster properties
        print("CRS:", src.crs)
        print("Shape:", src.shape)
        print("Resolution:", src.res)
        print("NoData:", src.nodata)
```
